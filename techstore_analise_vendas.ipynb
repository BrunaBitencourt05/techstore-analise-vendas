```python
import sqlite3
```

```python
df_vendas.shape[0]
```

```python
conn = sqlite3.connect("/content/vendas.csv")
```

```python
conn = sqlite3.connect(':memory:')

df_vendas.to_sql("vendas", conn, index = False, if_exists="replace")
```

```python
df_vendas = pd.read_sql_query("SELECT *, quantidade * preco_unitario AS receita_total FROM vendas", conn)
```

```python
df_vendas.head()
```

```python
#apenas eletrônicos
df_eletronicos = pd.read_sql_query("SELECT * FROM vendas WHERE categoria = 'Eletrônicos'", conn)
df_eletronicos.head()
```

```python
#mais vendido
df_mais_vendido = pd.read_sql_query("SELECT produto, SUM(quantidade) AS total_vendido FROM vendas GROUP BY produto ORDER BY total_vendido DESC LIMIT 1", conn)
df_mais_vendido.head()
```

```python
#região com o maior valor de compras
df_regiao = pd.read_sql_query("SELECT regiao, SUM(quantidade * preco_unitario) AS maior_valor FROM vendas GROUP BY regiao ORDER BY maior_valor DESC LIMIT 1", conn)
df_regiao.head()
```
